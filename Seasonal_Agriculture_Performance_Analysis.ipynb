###Handle Missing values
Choose and appropriate strategy based on the nature of each variable.

Possible approaches:


* Remove records only when justified
*   Fill numerical values using an appropriate statistic when appropriate

* Fill categorical values using a suitable category or method.   



```python
# Median imputation for columns with missing values

# Fill missing Rainfall values with the overall median
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(
    df['Rainfall_mm'].median()
)

# Fill missing Soil Moisture values with the overall median
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median()
)

# Fill missing Yield values with the overall median
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median()
)

# Check remaining missing values
print("Remaining missing values:", df.isnull().sum().sum())
```

###9. Outlier Analysis

Investigate potential outliers using statistical and visual methods

```python
# IQR based outlier summary for numerical columns

outlier_summary=[]
for col in numeric_col:
  Q1=df[col].quantile(0.25)
  Q3=df[col].quantile(0.75)
  IQR=Q3-Q1
  lower_bound=Q1-1.5*IQR
  upper_bound=Q3+1.5*IQR
  outliers=df[(df[col]<lower_bound) | (df[col]>upper_bound)]
  outlier_count=len(outliers)
  outlier_percentage=round((outlier_count/len(df))*100,2)
  outlier_summary.append([col,lower_bound,upper_bound,outlier_count,outlier_percentage])
outlier_summary=pd.DataFrame(outlier_summary,columns=['Column','lower_bound','upper_bound','Outlier Count','Outlier Percentage'])
outlier_summary
```

###13. Key Insights

###1. Kharif has the best overall performance

Observation:
Kharif performed better than Rabi and Zaid in terms of average yield and profit.

Evidence:
Average yield was around 5.63 tonnes/ha in Kharif, compared with 5.04 in Rabi and 4.64 in Zaid.

Interpretation:
Kharif appears to be the strongest season in this dataset for overall agricultural performance.

Limitation:
This result is based only on the farms and conditions included in the dataset. It does not mean Kharif will always be the best season everywhere.

###2. Yield changes significantly across seasons

Observation:
The average yield is not the same in all three seasons.

Evidence:
The Kruskal-Wallis test gave a p-value below 0.001, showing a significant difference in yield across seasons.

Interpretation:
Season is an important part of understanding the differences in agricultural performance in this dataset.

Limitation:
The test shows that the seasons are different, but it does not tell us exactly what caused the difference.

###3. Crop type has a major effect on yield

Observation:
Some crops have much higher yields than others.

Evidence:
Sugarcane had an average yield of around 46.64 tonnes/ha, while most other crops were around 1–3 tonnes/ha.

Interpretation:
Crop type needs to be considered when comparing agricultural performance instead of looking only at the overall average.

Limitation:
Different crops naturally have different yield levels, so their yields should not be compared without considering the type of crop.

###4. High yield does not always mean high profit

Observation:
A crop with lower yield can still make good profit.

Evidence:
Chilli had an average yield of around 1.54 tonnes/ha, but its average profit was around ₹7.5 lakh.

Interpretation:
Profit depends not only on yield but also on factors such as selling price, production cost and crop value.

Limitation:
The analysis shows a relationship between these factors but does not prove that one particular factor is responsible for higher profit.

###5. Water efficiency is different for different irrigation methods

Observation:
Water efficiency varies across irrigation methods.

Evidence:
Average water efficiency was approximately 7.56 for Rainfed, 6.27 for Drip, 4.67 for Sprinkler and 3.44 for Flood irrigation.

Interpretation:
Rainfed and drip irrigation show better water efficiency in this dataset.

Limitation:
The result can also be affected by rainfall, crop type, soil conditions and other farm characteristics.

###6. Kharif has higher disease and pest risk

Observation:
The recorded disease and pest risk is higher during Kharif.

Evidence:
Average risk was around 54.47% in Kharif, compared with 40.48% in Rabi and 38.22% in Zaid.

Interpretation:
Disease and pest management may need more attention during the Kharif season.

Limitation:
This column represents estimated risk, not the actual percentage of crops that were infected.

###7. Agricultural performance varies across states

Observation:
The average yield is different from one state to another.

Evidence:
For example, Punjab had an average yield of around 6.12 tonnes/ha, while Andhra Pradesh had around 4.63 tonnes/ha.

Interpretation:
Location can be an important area to investigate when trying to understand differences in agricultural performance.

Limitation:
State-level averages alone cannot explain why one state performs better than another.

###8. Seed quality does not show a clear relationship with yield

Observation:
Seed quality and yield have almost no linear relationship in this dataset.

Evidence:
The correlation between Seed_Quality_Score and Yield_Tonnes_Ha is approximately -0.01.

Interpretation:
Based on this dataset, having a higher seed quality score does not necessarily correspond to a higher yield.

Limitation:
This does not mean seed quality is unimportant. It only means that this dataset does not show a clear linear relationship between these two variables.

###9. Individual environmental factors do not strongly explain yield

Observation:
Rainfall, temperature, humidity and soil moisture do not individually show a strong relationship with yield.

Evidence:
Their correlations with yield are all close to zero in the correlation analysis.

Interpretation:
Yield appears to depend on a combination of different factors rather than one environmental factor alone.

Limitation:
Correlation only shows a linear relationship. Other types of relationships or differences between specific crops, seasons and regions may not be captured.

###14. Recommendation

Focus on Kharif: It performed better, so plan resources accordingly.

Choose crops carefully: Sugarcane showed good yield and profit in our data.

Manage water wisely: Use more efficient irrigation methods where suitable.

Watch for pests: Kharif had higher disease and pest risk, so regular monitoring is important.

Use data for planning: Compare seasons, crops, and regions before making decisions.

Keep the findings realistic: These are patterns from our dataset, not proof that one factor directly causes another.

###14. Conclusion

This analysis helped us understand how agricultural performance changes across seasons, crops, and regions. Kharif showed the best overall performance, while Zaid was comparatively lower. The seasonal differences in yield were also statistically significant.

The analysis showed that crop type plays an important role in overall performance, with Sugarcane having a much higher yield and profit than most of the other crops in the dataset. At the same time, profitability was not always directly related to yield, as some crops with lower yield still showed strong profit because of differences in market price and other economic factors

We also found differences across states and irrigation methods. Rainfed and drip irrigation showed better water efficiency, while Kharif had the highest disease and pest risk.

Rainfall, temperature, humidity, and seed quality did not show a strong direct relationship with yield. Overall, the results show that agricultural performance depends on multiple factors together and can help with better planning and resource management
